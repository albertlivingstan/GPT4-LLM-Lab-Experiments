# Experiment 12: Text Translation Assistant

**Aim:** Translate a paragraph into a target language and save translated_text.txt.

This notebook uses the local Ollama model `qwen2.5:3b`. No internet is needed while the model is installed and Ollama is running.

## 1. Connect to Ollama
Run this cell first.

```python
import requests

MODEL = "qwen2.5:3b"
OLLAMA_URL = "http://localhost:11434/api/generate"

def ask_ollama(prompt):
    response = requests.post(
        OLLAMA_URL,
        json={"model": MODEL, "prompt": prompt, "stream": False},
        timeout=300
    )
    response.raise_for_status()
    return response.json()["response"]

print("Ollama connection function is ready.")
print("Make sure Ollama is running and qwen2.5:3b is installed.")
```

## 2. Run the experiment
Enter the requested input when prompted.

```python
text = input("Enter text to translate: ")
language = input("Target language: ")
prompt = f"Translate the following text into {language}. Preserve its meaning. Return only the translation.\n{text}"
result = ask_ollama(prompt)
print(result)
with open("translated_text.txt", "w", encoding="utf-8") as f:
    f.write(result)
print("Saved as translated_text.txt")
```
