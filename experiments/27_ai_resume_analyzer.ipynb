# Experiment 27: AI Resume Analyzer

**Aim:** Read data/resume.txt and identify skills, qualifications, profile summary and improvements.

This notebook uses the local Ollama model `qwen2.5:3b`. No internet is needed while the model is installed and Ollama is running.

## 1. Connect to Ollama
Run this cell first.

```python
import requests

MODEL = "qwen2.5:3b"
OLLAMA_URL = "http://localhost:11434/api/generate"

def ask_ollama(prompt):
    response = requests.post(
        OLLAMA_URL,
        json={"model": MODEL, "prompt": prompt, "stream": False},
        timeout=300
    )
    response.raise_for_status()
    return response.json()["response"]

print("Ollama connection function is ready.")
print("Make sure Ollama is running and qwen2.5:3b is installed.")
```

## 2. Run the experiment
Enter the requested input when prompted.

```python
from pathlib import Path\nresume_path = Path("data/resume.txt") if Path("data/resume.txt").exists() else Path("../data/resume.txt")\nwith open(resume_path, "r", encoding="utf-8") as f:
    resume = f.read()
prompt = f"Analyze this resume. List skills, qualifications, a brief profile summary, and improvement recommendations. Do not make up facts.\n{resume}"
result = ask_ollama(prompt)
print(result)
```
