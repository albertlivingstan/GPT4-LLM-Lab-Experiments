# Experiment 8: College FAQ Chatbot

**Aim:** Answer questions using predefined college FAQ information and say when information is unavailable.

This notebook uses the local Ollama model `qwen2.5:3b`. No internet is needed while the model is installed and Ollama is running.

## 1. Connect to Ollama
Run this cell first.

```python
import requests

MODEL = "qwen2.5:3b"
OLLAMA_URL = "http://localhost:11434/api/generate"

def ask_ollama(prompt):
    response = requests.post(
        OLLAMA_URL,
        json={"model": MODEL, "prompt": prompt, "stream": False},
        timeout=300
    )
    response.raise_for_status()
    return response.json()["response"]

print("Ollama connection function is ready.")
print("Make sure Ollama is running and qwen2.5:3b is installed.")
```

## 2. Run the experiment
Enter the requested input when prompted.

```python
faq = """
Admission: Contact the college admission office for application details.
Courses: Refer to the official college prospectus for available courses.
Library: Library timings should be confirmed with the library office.
Examinations: Check official college notices for examination schedules.
"""
question = input("Ask a college FAQ: ")
prompt = f"Answer using ONLY the FAQ below. If the answer is not present, say 'Information not available in the FAQ.'\nFAQ:\n{faq}\nQuestion: {question}"
result = ask_ollama(prompt)
print(result)
```
