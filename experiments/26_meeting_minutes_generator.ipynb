# Experiment 26: Meeting Minutes Generator

**Aim:** Create summary, decisions and action items from notes; save meeting_minutes.txt.

This notebook uses the local Ollama model `qwen2.5:3b`. No internet is needed while the model is installed and Ollama is running.

## 1. Connect to Ollama
Run this cell first.

```python
import requests

MODEL = "qwen2.5:3b"
OLLAMA_URL = "http://localhost:11434/api/generate"

def ask_ollama(prompt):
    response = requests.post(
        OLLAMA_URL,
        json={"model": MODEL, "prompt": prompt, "stream": False},
        timeout=300
    )
    response.raise_for_status()
    return response.json()["response"]

print("Ollama connection function is ready.")
print("Make sure Ollama is running and qwen2.5:3b is installed.")
```

## 2. Run the experiment
Enter the requested input when prompted.

```python
notes = input("Enter meeting notes: ")
prompt = f"Prepare meeting minutes. Include summary, decisions, and action items/responsibilities explicitly mentioned. Do not invent details.\n{notes}"
result = ask_ollama(prompt)
print(result)
with open("meeting_minutes.txt", "w", encoding="utf-8") as f:
    f.write(result)
print("Saved as meeting_minutes.txt")
```
