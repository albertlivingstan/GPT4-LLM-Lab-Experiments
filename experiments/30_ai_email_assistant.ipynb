# Experiment 30: AI Email Assistant

**Aim:** Generate an email reply with subject, greeting, message and closing; save email_reply.txt.

This notebook uses the local Ollama model `qwen2.5:3b`. No internet is needed while the model is installed and Ollama is running.

## 1. Connect to Ollama
Run this cell first.

```python
import requests

MODEL = "qwen2.5:3b"
OLLAMA_URL = "http://localhost:11434/api/generate"

def ask_ollama(prompt):
    response = requests.post(
        OLLAMA_URL,
        json={"model": MODEL, "prompt": prompt, "stream": False},
        timeout=300
    )
    response.raise_for_status()
    return response.json()["response"]

print("Ollama connection function is ready.")
print("Make sure Ollama is running and qwen2.5:3b is installed.")
```

## 2. Run the experiment
Enter the requested input when prompted.

```python
email = input("Paste the original email: ")
tone = input("Tone (Formal/Friendly/Professional): ")
prompt = f"Draft an appropriate email reply in a {tone} tone. Include Subject, Greeting, Message, and Closing. Do not claim actions that were not provided.\nOriginal email:\n{email}"
result = ask_ollama(prompt)
print(result)
with open("email_reply.txt", "w", encoding="utf-8") as f:
    f.write(result)
print("Saved as email_reply.txt")
```
